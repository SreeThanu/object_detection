# Data Visualization

This notebook visualizes the VOC dataset samples.


```python
import sys
from pathlib import Path
import matplotlib.pyplot as plt

# Add parent directory to path
sys.path.append(str(Path().resolve().parent))

from src.visualization import visualize_dataset_samples
from src.preprocessing import VOCDataset, get_transforms
from config.config import RESULTS_DIR
```

## Visualize Training Samples


```python
# Create dataset
train_dataset = VOCDataset(split='train', transform=get_transforms(train=False))

# Visualize samples
save_dir = RESULTS_DIR / "visualizations" / "dataset_samples"
visualize_dataset_samples(train_dataset, num_samples=5, save_dir=save_dir)
```

## Visualize Validation Samples


```python
# Create validation dataset
val_dataset = VOCDataset(split='val', transform=get_transforms(train=False))

# Visualize samples
visualize_dataset_samples(val_dataset, num_samples=5, save_dir=save_dir)
```
