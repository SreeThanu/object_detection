# Model Training

This notebook trains the object detection model using ResNet-18 backbone.


```python
import sys
from pathlib import Path
import torch

# Add parent directory to path
sys.path.append(str(Path().resolve().parent))

from src.training import train_model, create_model
from config.config import TRAIN_CONFIG, MODEL_CONFIG
```

## Check Device


```python
# Check if CUDA is available
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
if torch.cuda.is_available():
    print(f"CUDA device: {torch.cuda.get_device_name(0)}")
    print(f"CUDA memory: {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")
```

## Create Model


```python
# Create model
model = create_model(
    num_classes=MODEL_CONFIG['num_classes'],
    pretrained=MODEL_CONFIG['pretrained']
)

print(f"Model created with {MODEL_CONFIG['num_classes']} classes")
print(f"Model architecture:")
print(model)
```

## Training Configuration


```python
# Display training configuration
print("Training Configuration:")
for key, value in TRAIN_CONFIG.items():
    print(f"  {key}: {value}")
```

## Start Training


```python
# Train model
# You can customize these parameters
results = train_model(
    model=model,
    num_epochs=TRAIN_CONFIG['num_epochs'],
    learning_rate=TRAIN_CONFIG['learning_rate'],
    batch_size=TRAIN_CONFIG['batch_size'],
    resume_from=None,  # Set to checkpoint path to resume training
    save_best=True
)
```

## Plot Training Curves


```python
# Plot training curves
from src.visualization import plot_training_curves
from config.config import RESULTS_DIR

plot_training_curves(
    results['train_losses'],
    results['val_losses'],
    save_path=RESULTS_DIR / "training_curves.png"
)
```

## Training Summary


```python
print(f"\nTraining completed!")
print(f"Best validation loss: {results['best_val_loss']:.4f}")
print(f"Final training loss: {results['train_losses'][-1]:.4f}")
print(f"Final validation loss: {results['val_losses'][-1]:.4f}")
```
